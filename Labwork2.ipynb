```python
import random

#Initial data
itemslist = ["Banana", "Strawberry", "Cherry", "Chocolate", "Milk", "Egg", "Cola", "Pepsi", "Juice", "Bread", "Cake"]
shoplist = ["Awesomest", "fourerr", "Bluu", "Competitor", "Chippy"]
shoplistprices = ["Awesomest", "fourerr", "Bluu", "Competitor", "Chippy"]
shoptotal = ["Awesomest", "fourerr", "Bluu", "Competitor", "Chippy"]
shopresult = []

#Just a function for ease of use
def generate_prices():
  return random.randint(1,100)

#Creating costs for every shop
for x in range(len(shoplist)):
  shoplistprices[x] = []
  for y in range(len(itemslist)):
    shoplistprices[x] = shoplistprices[x] + [(itemslist[y], generate_prices())]
  print(shoplist[x], shoplistprices[x])

#Persons shoppinglist
print("Please type shopping list below")
shoppinglist = input().split(", ")

def count_total_sum(shop_name, product_list):
  total=0
  break_notice=0
  for z in range(len(product_list)):
    for x in range(len(shoplist)):
      if break_notice==len(product_list):
        break
      elif shoplist[x].lower() == shop_name.lower():
        for y in range(len(itemslist)):
          if shoplistprices[x][y][0].lower() == product_list[z].lower():
            total=total+shoplistprices[x][y][1]
            break_notice += 1
            break
  return total

for x in range(len(shoplist)):
  shoptotal[x]=count_total_sum(shoplist[x], shoppinglist)
  shopresult = shopresult + [(shoptotal[x], shoplist[x])]

print(shopresult)
print("best store to buy is from " + sorted(shopresult)[0][1] + " and the cost is " + str(sorted(shopresult)[0][0]))
```
